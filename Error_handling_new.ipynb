```python
revenue = int(input("Enter the revenue: "))
expenses = int(input("Enter the expenses: "))
```

```python
def safe_margin_calculation():
    try:
        profit = revenue - expenses
        print(f"Profit: {profit}")   

        margin = (profit / revenue) * 100
        print(f"Profit Margin: {margin:.2f}%") 
    except ValueError:
        print("Invalid input. Please enter numeric values for revenue and expenses.")
    except ZeroDivisionError:
        print("Revenue cannot be zero when calculating profit margin.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
safe_margin_calculation()
```
