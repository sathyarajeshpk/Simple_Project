# **Functions**

```python
import datetime

def calculate_age(birth_year):
    current_year = datetime.datetime.now().year
    age = current_year - birth_year
    return (age)

current_year = datetime.datetime.now().year
while True:
    try:
        birth_year = int(input("Enter your birth year (e.g., 1990): "))
        if birth_year <= current_year and birth_year > current_year - 120: # Assuming a realistic age range (e.g., not older than 120)
            break
        else:
            print(f"Please enter a realistic birth year between {current_year - 120} and {current_year}.")
    except ValueError:
        print("Invalid input. Please enter a numerical year.")

age = calculate_age(birth_year)
print(f"Your age is {age} years.")
```

```python
class Human:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def average_age(self, other):
        return (self.age + other.age) / 2

sathya = Human("Sathya", 36, "Male")
kirthiga = Human("Kirthiga", 35, "Female")

print(sathya.average_age(kirthiga))
```

```python
people = [
    Human("Sathya", 36, "Male"),
    Human("Kirthiga", 35, "Female"),
    Human("John", 15, "Male"),
    Human("Jane", 28, "Female"),
    Human("Alice", 41, "Female"),
    Human("Bob", 25, "Male"),
    Human("Eve", 32, "Female")
]

print("List of people created successfully.")
```

```python
def calculate_average_age_of_group(human_list):
    if not human_list:
        return 0
    total_age = sum(human.age for human in human_list)
    return total_age / len(human_list)

def calculate_average_gender(human_list):
    if not human_list:
        return 0
    male_count = sum(1 for human in human_list if human.gender == "Male")
    female_count = sum(1 for human in human_list if human.gender == "Female")
    return male_count / female_count

average_group_age = calculate_average_age_of_group(people)
average_gender = calculate_average_gender(people)

if average_group_age == 0:
    print("No people in the group.")
else:

  print(f"The average age of the group is: {average_group_age:.2f} years")
  print(f"The average gender of the group is: {average_gender:.2f}")
```
