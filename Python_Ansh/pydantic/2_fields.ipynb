## Fields

```python
from pydantic import BaseModel, Field, StrictStr, StrictInt, EmailStr
from typing import Optional, Literal
```

```python
class employee_schema(BaseModel):
    name: StrictStr = Field(..., title="Employee Name", description="Name of the employee", max_length=50)
    age: StrictInt = Field(..., title="Employee Age", description="Age of the employee", ge=18, le=65)
    email: Optional[EmailStr] = Field(None, title="Employee Email", description="Email address of the employee")
    department: Literal["HR", "IT", "Finance", "Marketing"] = Field(..., title="Employee Department", description="Department of the employee")
```

```python
obj = employee_schema(**{"name": "Ansh", "age": 22, "email": "ansh@example.com", "department": "IT"})
obj
```
