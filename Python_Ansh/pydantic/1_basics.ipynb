```python
from pydantic import BaseModel, StrictStr, StrictInt
```

```python
class employee_schema(BaseModel):
    name: str
    age: int
    salary: float
    department: str
    gender: str
```

```python
obj = employee_schema(name="Ansh", age=22, salary=50000.0, department="IT", gender="Male")
print(obj)
```

```python
obj2 = employee_schema(**{"name": "Ansh", "age": "22", "salary": 50000.0, "department": "101", "gender": "Male"})
print(obj2)
```

```python
class employee_schema(BaseModel):
    name: StrictStr
    age: StrictInt
    salary: float
    department: str
    gender: str
```

```python
obj3 = employee_schema(**{"name": "Ansh", "age": 22, "salary": 50000.0, "department": "101", "gender": "Male"})
print(obj3)
```
