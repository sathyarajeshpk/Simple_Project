# **Calling LLM**

**Practising LLM calling from VS Code using Python**

```python
from dotenv import load_dotenv
from google import genai
```

```python
try:
    if not load_dotenv():
        raise RuntimeError(".env file could not be loaded.")
except Exception as e:
    print(f"{e}. Please ensure the file exists and is properly formatted. ")
```

```python
# Initialize the client (automatically captures GEMINI_API_KEY from environment)
client = genai.Client()


# Generate text content using a Gemini model
try:
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents="Write a short story about a robot learning to love."
    )
    print(response.text)
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
def generate_recipe_errhandle(ingredients, cuisine):
    
    prompt = f"Write a {', '.join(cuisine)} recipe using the following ingredients: {', '.join(ingredients)}"
    
    try:
        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=prompt
        )
        if not response.text:
            raise ValueError("Received empty response from the model.")
        return response.text
    except Exception as e:
        print(f"An error occurred: {e}")
        return "Sorry, I couldn't generate a recipe at the moment."
    
```

```python
# recipe = generate_recipe(["chicken", "garlic", "onion", "tomato"], ["Indian", "Italian"])
recipe = generate_recipe_errhandle(["chicken", "garlic", "onion", "tomato"], ["Indian", "Italian"])
print(recipe)
```
